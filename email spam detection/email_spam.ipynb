Email Spam Detection using Logistic Regression



First let's load the csv file and visualise our dataset

```python
import pandas as pd
df = pd.read_csv('email_spam_dataset_250.csv')
df.head()
```

```python
df.shape #-> (rows,columns)
```

```python
df.columns
```

```python
df['label'].value_counts()
```

converting the spam:1 and ham:0

```python
df['label'] = df['label'].map({'ham':0,'spam':1})
```

```python
df.head()
```

Let's vectorize the text to be a proper input

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer(stop_words='english',max_features=500)
X=tfidf.fit_transform(df['text'])
Y=df['label']
print(X.shape)
print(Y.shape)
```

```python
feature_names=tfidf.get_feature_names_out()
print(feature_names)
```

```python
test=X.toarray()
for i in range(len(test[0])):
  print(feature_names[i],test[0][i])
```

Let's split the data in 80-20 set

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
print(X_train.shape)
print(X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
model =LogisticRegression()
model.fit(X_train,Y_train)
```

```python
print(accuracy_score(Y_test,model.predict(X_test)))
```
